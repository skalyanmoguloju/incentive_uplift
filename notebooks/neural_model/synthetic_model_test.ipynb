# Synthetic Data Model Test

This notebook tests the expected inputs and outputs of the multi-treatment uplift model using synthetic data. The synthetic data allows the neural-model pipeline to be developed before the final processed dataset is available.

```python
import torch

# Setting this seed to fix the randomness
torch.manual_seed(42)

n_records = 64 # Number of sample records for creating the data
n_customer_features = 8 # An arbitory number defing the features available for each customer
n_incentive_features = 6 # Features for each offer or incentive
n_treatments = 11 # Number of treatments this will be including the control treatment as well

# Creating customer dataset
customer_features = torch.randn(
    n_records,
    n_customer_features
)

# Creating incentive dataset
incentive_features = torch.randn(
    n_records,
    n_incentive_features
)

# Creating treatment dataset for the sample data
treatment = torch.randint(
    low=0,
    high=n_treatments,
    size=(n_records,)
)

# Creating outcome label for the sample data
outcome = torch.randint(
    low=0,
    high=2,
    size=(n_records, 1)
).float()


print("Customer features:", customer_features.shape)
print("Incentive features:", incentive_features.shape)
print("Treatment labels:", treatment.shape)
print("Outcomes:", outcome.shape)
```

## Synthetic Data Validation


```python
# Validation to make sure the treament assignment is done as expected
print("Treatment minimum:", treatment.min().item())
print("Treatment maximum:", treatment.max().item())
print("Unique treatments:", torch.unique(treatment).tolist())

# Validating the outcomes is binary
print("Unique outcomes:", torch.unique(outcome).tolist())

# Making sure no feature data is invalid
print("Customer features contain NaN:",
      torch.isnan(customer_features).any().item())
print("Incentive features contain NaN:",
      torch.isnan(incentive_features).any().item())
print("Outcomes contain NaN:",
      torch.isnan(outcome).any().item())

# Data assertions to make sure all the datasets are in valid state
assert treatment.min().item() >= 0
assert treatment.max().item() < n_treatments
assert set(torch.unique(outcome).tolist()).issubset({0.0, 1.0})
assert not torch.isnan(customer_features).any().item()
assert not torch.isnan(incentive_features).any().item()
assert not torch.isnan(outcome).any().item()

print("Synthetic data validation is completed.")
```
