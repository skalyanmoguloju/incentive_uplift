# Data loader testing using the generated data

```python
import torch

# Setting this seed to fix the randomness
torch.manual_seed(42)

n_records = 64 # Number of sample records for creating the data
n_customer_features = 8 # An arbitory number defing the features available for each customer
n_incentive_features = 6 # Features for each offer or incentive
n_treatments = 11 # Number of treatments this will be including the control treatment as well

# Creating customer dataset
customer_features = torch.randn(
    n_records,
    n_customer_features
)

# Creating incentive dataset
incentive_features = torch.randn(
    n_records,
    n_incentive_features
)

# Creating treatment dataset for the sample data
treatment = torch.randint(
    low=0,
    high=n_treatments,
    size=(n_records,)
)

# Creating outcome label for the sample data
outcome = torch.randint(
    low=0,
    high=2,
    size=(n_records, 1)
).float()
```

```python
from torch.utils.data import Dataset, DataLoader

```

```python
# Wrapper class for grouping all the tensors on a dataset
class SyntheticUpliftDataset(Dataset):
    def __init__(
        self,
        customer_features,
        incentive_features,
        treatment,
        outcome
    ):
        self.customer_features = customer_features
        self.incentive_features = incentive_features
        self.treatment = treatment
        self.outcome = outcome

    # To know how many samples the dataset include
    def __len__(self):
        return self.customer_features.shape[0]

    # Returns a single dataset based on the index used for single customer and treatment decision
    def __getitem__(self, index):
        return {
            "customer_features": self.customer_features[index],
            "incentive_features": self.incentive_features[index],
            "treatment": self.treatment[index],
            "outcome": self.outcome[index]
        }
```

```python
# Created the dataset instance using the generated data
dataset = SyntheticUpliftDataset(
    customer_features=customer_features,
    incentive_features=incentive_features,
    treatment=treatment,
    outcome=outcome
)

print("Dataset size:", len(dataset))


# Sample dataset validation
sample = dataset[0]

print("Sample keys:", sample.keys())
print("Customer features:", sample["customer_features"].shape)
print("Incentive features:", sample["incentive_features"].shape)
print("Treatment:", sample["treatment"].shape)
print("Outcome:", sample["outcome"].shape)
```

```python
# Loader loading 8 dataset at a time in a shuffle mode
data_loader = DataLoader(
    dataset,
    batch_size=8,
    shuffle=True
)


# Fetching the first batch and validating the shapes for each data set
batch = next(iter(data_loader))

print("Customer features:", batch["customer_features"].shape)
print("Incentive features:", batch["incentive_features"].shape)
print("Treatment:", batch["treatment"].shape)
print("Outcome:", batch["outcome"].shape)
```
